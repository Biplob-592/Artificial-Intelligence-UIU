Required Tasks :

b. Data Cleaning:
1.If you find any missing values in the dataset (nan values) replace those with
the column-wise median for numeric data, and with mode for categorical
data.
2.If you find any duplicates in the dataset, keep just one copy of the data.
3.Remember, if any row in the target column (Attrition) is missing, you must
drop that row.
4.Remove any outliers present in the numeric columns.
5.Standardize the string columns data to lower case.


```python
#b. 1. If you find any missing values in the dataset (nan values) replace those with
#      the column-wise median for numeric data, and with mode for categorical data.

numeric_cols = df.select_dtypes(include=[np.number]).columns # Select numeric columns
categorical_cols = df.select_dtypes(include=['object']).columns # Select categorical columns
```

```python
# Fill missing values in numeric columns with median
for col in numeric_cols:
    df[col]= df[col].fillna(df[col].median())
df
```

```python
# Fill missing values in categorical columns with mode
for col in categorical_cols:  
    df[col] = df[col].fillna(df[col].mode()[0])
df
```

```python
#b.2. If you find any duplicates in the dataset, keep just one copy of the data.
df = df.drop_duplicates()
df
```

```python
#b.3 Remember, if any row in the target column (Attrition) is missing, you must drop that row.

df.dropna(subset=['Attrition'], inplace=True) # Drop rows where 'Attrition' is missing
df
```

```python
#b.4. Remove any outliers present in the numeric columns.
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import zscore
```

```python
#finding outliers using boxplot
for col in numeric_cols:

    plt.figure(figsize=(12,6))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
z_threshold = 3

# Only use columns that have more than one unique value
varying_cols = [col for col in numeric_cols if df[col].nunique() > 1]

z_scores = np.abs(zscore(df[varying_cols]))
outlier_mask = (z_scores < z_threshold).all(axis=1)

df = df[outlier_mask]
df
```

```python
#After removing outliers, we can visualize the boxplots again to confirm that the outliers have been removed.
for col in numeric_cols:

    plt.figure(figsize=(12,6))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
#b.5. Standardize the string columns data to lower case.
for col in categorical_cols:
    df[col] = df[col].str.lower()
df
```
