```python
import heapq
grid = [
['S','.','.','.','#','.','.','.','.','.'],
['.','#','#','.','#','.','#','#','#','.'],
['.','.','.','.','.','.','.','.','#','.'],
['#','#','.','#','#','#','.','#','#','.'],
['.','.','.','.','.','#','.','.','.','.'],
['.','#','#','#','.','#','#','#','.','#'],
['.','.','.','#','.','.','.','#','.','.'],
['#','#','.','#','#','#','.','#','#','.'],
['.','.','.','.','.','.','.','.','#','.'],
['.','#','#','#','#','#','.','.','.','G']
]

rows = len(grid)
cols = len(grid[0])

dirs = [(-1,0),(1,0),(0,-1),(0,1)]

# Finding start and goal
def find_start_goal():
    for i in range(rows):
        for j in range(cols):
            if grid[i][j] == 'S':
                start = (i, j)
            if grid[i][j] == 'G':
                goal = (i, j)
    return start, goal

# Using Manhattan heuristic (As given & discussed in class)
def heuristic(a, b):
    return abs(a[0] - b[0]) + abs(a[1] - b[1])


# Function for A* and dijkstra
def shortest_path(start, goal, mode):

    pq = []
    heapq.heappush(pq, (0, start))

    dist = {start: 0}
    parent = {}
    visited = 0

    while pq:
        visited += 1
        priority, node = heapq.heappop(pq)

        if node == goal:
            break

        for d in dirs:
            nx = node[0] + d[0]   #nx= new x-coordinate
            ny = node[1] + d[1]   #ny = new Y-coordinate

            if 0 <= nx < rows and 0 <= ny < cols:
                if grid[nx][ny] != '#':

                    new_cost = dist[node] + 1

                    if (nx, ny) not in dist or new_cost < dist[(nx, ny)]:
                        dist[(nx, ny)] = new_cost
                        parent[(nx, ny)] = node

                        if mode == 'dijkstra':
                            priority = new_cost
                        elif mode == 'astar':
                            priority = new_cost + heuristic((nx, ny), goal)

                        heapq.heappush(pq, (priority, (nx, ny)))

    # Path reconstruction
    path = []
    node = goal

    while node != start:
        path.append(node)
        node = parent[node]

    path.append(start)
    path.reverse()

    return visited, path, dist[goal]


# Main
if __name__ == "__main__":
    start, goal = find_start_goal()

    # Dijkstra
    vis1, path1, cost1 = shortest_path(start, goal, mode="dijkstra")

    print("DIJKSTRA RESULT:")
    print("Path:", path1)
    print("Cost:", cost1)
    print("Nodes explored:", vis1)

    print()

    # A*
    vis2, path2, cost2 = shortest_path(start, goal, mode="astar")

    print("A* RESULT:")
    print("Path:", path2)
    print("Cost:", cost2)
    print("Nodes explored:", vis2)
```
